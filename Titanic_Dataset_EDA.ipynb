```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the Titanic dataset
df = pd.read_csv("https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv")
```

## Missing Values Treatment

```python
df.isnull().sum()
```

```python
df.isnull().sum()/len(df)*100
```

```python
df.drop('Cabin',axis = 1, inplace = True)
```

```python
df
```

```python
df.Age
```

```python
df.Age.fillna(df.Age.median(),inplace = True)
```

```python
df.Embarked.fillna(df.Embarked.mode()[0],inplace=True)
```

```python
df.isnull().sum()
```

```python
sns.countplot(x = 'Survived',data = df,hue='Survived', palette='viridis')
plt.show()
```

```python
sns.pairplot(df[['Survived','Age','Pclass','Fare']])
```

```python
sns.histplot(df['Age'],bins = 20,kde = True)
plt.show()
```

```python
sns.boxplot(x = 'Pclass', y = 'Age', data = df)
plt.title("Age by Passenger Class")
plt.show()
```

```python
sns.countplot(x ='Sex', hue = 'Pclass', data = df,palette = 'coolwarm')
plt.show()                                                                                #Bivariate Analysis
```

```python
sns.countplot(x='Sex', hue='Survived', data=df,palette='viridis')
plt.title("Survival by Embarked Port")
plt.show()
```

```python
sns.heatmap(df.corr(numeric_only=True),cmap = 'coolwarm', annot  =True)
plt.title('Correlation Heatmap')
plt.show()
```

```python
sns.scatterplot(x = 'Age', y = 'Fare',hue = 'Survived', data = df)
```
